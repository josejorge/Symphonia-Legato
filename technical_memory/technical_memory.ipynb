## 2026-09-15 — Repo status audit + TODO.md triage kickoff

**Context.** The project owner believed the app "never worked." A fresh audit found the
opposite: `dotnet build SymphoniaLegato.sln` succeeds with 0 warnings/0 errors, 111/111
tests pass, and the Desktop `.exe` launches to a real, responsive window.

**Root cause of the false impression.** `run_err.txt` in the repo root was a *stale*
startup-crash log (`InputGesture="Plus"` is not a valid Avalonia key name — throws
`ArgumentException` on `MainWindow` XAML load) predating commit `3d64da4`. The current
`MainWindow.axaml` already uses `OemPlus`/`OemMinus`/`D0`. `docs/BUGFIXES.md` already
documented this file as stale ("Notes / non-bugs" section) — it just never got deleted.

**Decision: delete the stale debug artifacts instead of investigating them further.**
`msbuild.binlog` (428 KB) and `run_err[23].txt` / `run_out[23].txt` were committed to
the repo root. `docs/TODO.md` already listed removing them as a known item. Untracked
via `git rm --cached` + deleted from disk, and added `*.binlog` / `run_out*.txt` /
`run_err*.txt` to `.gitignore` so they can't be re-committed by accident.

**Decision: `docs/TODO.md`'s backlog was triaged into tiers, not rewritten as a new
plan document.** Tiering rationale: Tier 0 (repo hygiene, no design decisions) → Tier 1
(small correctness bugs, no design decisions) → Tier 2 (core editor completeness —
what most affects daily usability) → Tier 3 (depth/polish features) → Tier 4
(long-term/major undertakings: real-time MIDI input, collaborative editing, Android
editing parity). The user picked Tier 0 + Tier 1 to start.

**Note for future sessions:** the empty `tests/SymphoniaLegato.PlaybackEngine.Tests`
project (wired into the solution, zero test files) was found during the audit — it
silently contributes 0 of the tests implied by the project's own "111 tests" milestone
claim in the root `CLAUDE.md` phase table. See CHANGELOG.md for what was actually done
about it this session.
